```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401 (needed for 3D)

# Sample resolution
n = 300  # increase/decrease for density
u = np.linspace(0, 2*np.pi, n)
v = np.linspace(0, 2*np.pi, n)
U, V = np.meshgrid(u, v)

# Clifford torus in S^3: (x1,x2,x3,x4) = (cos u, sin u, cos v, sin v) / sqrt(2)
s = 1/np.sqrt(2)
x1 = s*np.cos(U)
x2 = s*np.sin(U)
x3 = s*np.cos(V)
x4 = s*np.sin(V)

# Stereographic projection from S^3 \ {north pole (0,0,0,1)} to R^3:
# (X,Y,Z) = (x1, x2, x3) / (1 - x4)
denom = 1 - x4
X = x1 / denom
Y = x2 / denom
Z = x3 / denom

# Color by magnitude (distance from origin after projection)
mag = np.sqrt(X**2 + Y**2 + Z**2)

# Plot
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')

# Use matplotlib's default colormap; do not set colors explicitly
sc = ax.scatter(X.ravel(), Y.ravel(), Z.ravel(),
                c=mag.ravel(), s=2, depthshade=False)

cb = plt.colorbar(sc, ax=ax, fraction=0.03, pad=0.05)
cb.set_label("‖(X,Y,Z)‖")

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.set_title("Points on a Clifford Torus (stereographic projection) — colored by magnitude")

# Make aspect roughly equal
max_range = np.array([X.max()-X.min(), Y.max()-Y.min(), Z.max()-Z.min()]).max()
mid_x = (X.max()+X.min())*0.5
mid_y = (Y.max()+Y.min())*0.5
mid_z = (Z.max()+Z.min())*0.5
ax.set_xlim(mid_x - max_range/2, mid_x + max_range/2)
ax.set_ylim(mid_y - max_range/2, mid_y + max_range/2)
ax.set_zlim(mid_z - max_range/2, mid_z + max_range/2)
ax.set_box_aspect((1,1,1))

plt.tight_layout()
plt.show()
```
